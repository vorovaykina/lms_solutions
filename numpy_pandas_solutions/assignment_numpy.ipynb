* __Задача 1__: Подсчитать произведение ненулевых элементов на диагонали прямоугольной матрицы.  
 Например, для X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) ответ 3.

```python
import numpy as np
X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]])
diag = np.diag(X)
print(np.prod(diag[diag != 0]))

l = 1
for i in range(min(len(X), len(X[0]))):
    if X[i][i] != 0:
        l *= X[i][i]
print(l)
```

* __Задача 4__: Реализовать кодирование длин серий (Run-length encoding). Для некоторого вектора x необходимо вернуть кортеж из двух векторов одинаковой длины. Первый содержит числа, а второй - сколько раз их нужно повторить.  
 Например, для x = np.array([2, 2, 2, 3, 3, 3, 5]) ответ (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
import numpy as np
x = np.array([2, 2, 2, 3, 3, 3, 5])

ind = np.where(np.diff(x) != 0)[0]
ind = np.concatenate([[-1], ind, [len(x) - 1]])

count = np.diff(ind)
value = x[ind[1:]]
print((value, count))

x = [2, 2, 2, 3, 3, 3, 5]
value = []
count = []
if x:
    curr_value = x[0]
    curr_count = 0
    for val in x:
        if val == curr_value:
            curr_count += 1
        else:
            value.append(curr_value)
            count.append(curr_count)
            curr_value = val
            curr_count = 1
    value.append(curr_value)
    count.append(curr_count)
print((value, count)) 
```

* __Задача 5__: Даны две выборки объектов - X и Y. Вычислить матрицу евклидовых расстояний между объектами. Сравните с функцией scipy.spatial.distance.cdist по скорости работы.

```python
import numpy as np
from scipy.spatial.distance import cdist
import time

X = np.random.random((100, 10))
Y = np.random.random((100, 10))

def dist_(X, Y):
    return np.sqrt(np.sum(X**2, axis=1)[:, np.newaxis] + np.sum(Y**2, axis=1) - 2 * np.dot(X, Y.T))

st1 = time.time()
res1 = dist_(X, Y)
print(f"numpy time: {time.time() - st1}")

st2 = time.time()
res2 = cdist(X, Y)
print(f"scipy time: {time.time() - st2}")

matrix = []
for x_i in X:
    row = []
    for y_i in Y:
        d = sum((a - b) ** 2 for a, b in zip(x_i, y_i)) ** 0.5
        row.append(d)
    matrix.append(row)
print(matrix)
```
